# Section 1: List creation and accessing

```python
# Câu 1: Tạo mới 1 list gồm các số từ 1 đến 10
list_1 = list(range(1,11))
print(list_1)
```

```python
# in ra 5 phần tử đầu tiên có trong list trên.
print(list_1[:5])
```

```python
# in ra kết quả các phần tử có giá trị không chia hết cho 2( sử dụng kết hợp với vòng lặp for)
result =[]
for i in list_1:
    if i % 2 != 0:
      result.append(i)
print(result)
```

```python
# In ra kết quả tổng các giá trị trong list( dùng kết hợp với vòng lặp for)
list_sum = 0
for i in list_1:
  list_sum += i
print(list_sum)
```

Giới thiệu về số Armstrong: Số Armstrong(còn được gọi là số Narcissistic) là một số nguyên dương có giá trị bằng tổng lập phương của cá chữ số cấu thành nó.

```python
# Kiểm tra số armstrong 1 list cho trước
```

```python
# get the all digits of an interger

def get_digits(n):
    return [int(digit) for digit in str(n)]
get_digits(1235)
```

```python
# Tinh lap phuong cua tung chu so
def sum_cube(n):
  digits_lst = [int(digit) for digit in str(n)]
  print(digits_lst)
  sum_cube = 0
  for i in digits_lst:
    sum_cube += i**3
  return sum_cube
sum_cube(1634)
```

```python
test_case = [130, 270, 153, 407, 177, 371, 1000, 1634, 370]
```

```python
def check_armstrong(n):
  if n == sum_cube(n):
    return 1
  else:
    return 0
check_armstrong(1634)
```

```python
lst_armstrong = []
for i in test_case:
   if check_armstrong(i) == 1:
    lst_armstrong.append(i)
   # print(f' Kết quả check số armstrong là: {lst_armstrong}')
# result = test_case[lst_armstrong]
print(f'Các số Armstrong có trong list là: {lst_armstrong}')
```
